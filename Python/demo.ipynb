```python
class computer:
    def __init__(self):
        self.name = "PC"
        self.ram = 16
        self.cpu = "Intel i5"
    
    def update(self):
        self.age = 30
    
    def compare(self, other):
        if self.ram > other.ram:
            return f"{self.name} has more RAM than {other.name}"
        elif self.ram < other.ram:
            return f"{self.name} has less RAM than {other.name}"
        else:
            return f"{self.name} and {other.name} have the same amount of RAM"
        
```
