```python
df.loc[(df["age"]<18) | (df["age"]>100), "age"] = np.nan
df.loc[(df["tenure_months"]<0) | (df["tenure_months"]>72), "tenure_months"] = np.nan
df.loc[(df["monthly_charges"]<0) | (df["monthly_charges"]>200), "monthly_charges"] = np.nan
df.loc[(df["support_tickets_90d"] < 0), "support_tickets_90d"] = np.nan
```

```python
categorical_cols = df.select_dtypes(include="object").columns

for col in categorical_cols:
    df[col] = df[col].str.strip().str.lower()
```

```python
df["gender"] = df["gender"].replace({
    "m": "male",
    "f": "female"
})

df["contract_type"] = df["contract_type"].replace({
    "month to month": "month-to-month",
    "1 year": "one year",
    "one-year": "one year",
    "2 years": "two year",
    "two-year": "two year"
})

df["payment_method"] = df["payment_method"].replace({
    "e-check": "electronic check",
    "cc": "credit card",
    "transfer": "bank transfer",
    "paper check": "mailed check"
})

df["paperless_billing"] = df["paperless_billing"].replace({
    "yes": "yes",
    "no": "no"
})

df["city_tier"] = df["city_tier"].replace({
    "tier-1": "tier 1",
    "tier-2": "tier 2",
    "tier-3": "tier 3"
})
```

```python
df.replace(
    ["?", "n/a", "", "unknown"],
    np.nan,
    inplace=True
)
```

```python
for col in categorical_cols:
    print(f"\n{col}")
    print(df[col].value_counts(dropna=False))
```
